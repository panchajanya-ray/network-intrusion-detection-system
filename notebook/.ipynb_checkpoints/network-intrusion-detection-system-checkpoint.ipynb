```python
import numpy as np
import pandas as pd
```

# 1. Loaded Train Data

```python
# List of all 43 columns in the NSL-KDD dataset
columns = ["duration","protocol_type","service","flag","src_bytes","dst_bytes","land",
           "wrong_fragment","urgent","hot","num_failed_logins","logged_in",
           "num_compromised","root_shell","su_attempted","num_root","num_file_creations",
           "num_shells","num_access_files","num_outbound_cmds","is_host_login",
           "is_guest_login","count","srv_count","serror_rate","srv_serror_rate",
           "rerror_rate","srv_rerror_rate","same_srv_rate","diff_srv_rate",
           "srv_diff_host_rate","dst_host_count","dst_host_srv_count",
           "dst_host_same_srv_rate","dst_host_diff_srv_rate","dst_host_same_src_port_rate",
           "dst_host_srv_diff_host_rate","dst_host_serror_rate","dst_host_srv_serror_rate",
           "dst_host_rerror_rate","dst_host_srv_rerror_rate","label","difficulty"]

# Load the training data
df = pd.read_csv('../data/KDDTrain+.txt', header=None, names=columns)

# Display the first 5 rows and the attack types
print("Data Loaded Successfully!")
print(df['label'].value_counts().head()) # Shows the top 5 traffic types (normal vs attacks)
```

# 2. Data Preprocessing

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
# 1. Drop the 'difficulty' column (It's not a network feature, so we don't need it)
if 'difficulty' in df.columns:
    df.drop('difficulty', axis=1, inplace=True)

# 2. Binary Classification: Convert labels to 0 (Normal) and 1 (Attack)
df['label'] = df['label'].apply(lambda x: 0 if x == 'normal' else 1)

# 3. One-Hot Encoding: Convert text (tcp, http, etc.) into numbers
# This creates new columns for each category
df = pd.get_dummies(df, columns=['protocol_type', 'service', 'flag'])

# 4. Split data into Features (X) and Target (y)
X = df.drop('label', axis=1)
y = df['label']

# 5. Feature Scaling: Bring all values between 0 and 1
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for easier viewing
X_final = pd.DataFrame(X_scaled, columns=X.columns)

print("Preprocessing Complete!")
print(f"New feature count: {X_final.shape[1]} columns")
print(X_final.head())
```

# 3. Spliting Data

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_final, y, test_size=0.2, random_state=42)

print(f"Training set size: {X_train.shape[0]}")
print(f"Testing set size: {X_test.shape[0]}")
```

# 4. Building Model

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_train, y_train)

y_pred = rf_classifier.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
classification_rep = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy:.2f}")
print("\nClassification Report:\n", classification_rep)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Normal', 'Attack'], yticklabels=['Normal', 'Attack'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix - Network Intrusion Detection')
plt.show()
```

# 5. Testing The Model (KDDTest+.txt)

```python
# Load the new test dataset
df_test = pd.read_csv('KDDTest+.txt', header=None, names=columns)

# 1. Drop the difficulty column immediately
df_test.drop('difficulty', axis=1, inplace=True)

# 2. Binary labels (Normal vs Attack)
df_test['label'] = df_test['label'].apply(lambda x: 0 if x == 'normal' else 1)

# 3. One-Hot Encoding
df_test = pd.get_dummies(df_test, columns=['protocol_type', 'service', 'flag'])

# IMPORTANT: The test set might have DIFFERENT services than the train set.
# We must make sure the columns match our trained model exactly.
# This code adds missing columns and removes extra ones.
X_unseen = df_test.drop('label', axis=1)
y_unseen = df_test['label']

# Match columns of test set to training set (X_final from your previous step)
X_unseen = X_unseen.reindex(columns=X_final.columns, fill_value=0)

# 4. Scale using the SAME scaler from before
X_unseen_scaled = scaler.transform(X_unseen) 

print("New Test Set Loaded and Matched!")
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Convert the scaled array back to a DataFrame with original feature names
X_unseen_df = pd.DataFrame(X_unseen_scaled, columns=X_final.columns)

# Now predict using the DataFrame
y_unseen_pred = rf_classifier.predict(X_unseen_df)

# Final Results
print(f"Realistic Accuracy: {accuracy_score(y_unseen, y_unseen_pred):.2f}")
print("\nNew Attack Classification Report:\n", classification_report(y_unseen, y_unseen_pred))
```

# 6. Using XGBoost To Improve The Accuracy

```python
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report

# 1. Initialize the XGBoost Classifier
# We use 'scale_pos_weight' to help the model focus more on the attacks (Class 1)
xgb_model = xgb.XGBClassifier(
    n_estimators=200, 
    learning_rate=0.1, 
    max_depth=6, 
    verbosity=1,
    eval_metric='logloss' # Removed 'use_label_encoder'
)

# 2. Train the model
xgb_model.fit(X_train, y_train)

# 3. Predict on the "Real World" Unseen Test Set (X_unseen_df from earlier)
y_unseen_pred_xgb = xgb_model.predict(X_unseen_df)

# 4. Check results
print(f"XGBoost Realistic Accuracy: {accuracy_score(y_unseen, y_unseen_pred_xgb):.2f}")
print("\nXGBoost Classification Report:\n", classification_report(y_unseen, y_unseen_pred_xgb))
```

# 7. Exporting The XGBoost Model

```python
import pickle

# Save the trained XGBoost model
with open('intrusion_model.pkl', 'wb') as f:
    pickle.dump(xgb_model, f)

# Save the scaler (Crucial: the UI must scale data the exact same way)
with open('scaler.pkl', 'wb') as f:
    pickle.dump(scaler, f)
    
# Save the column names so the UI knows the order
with open('columns.pkl', 'wb') as f:
    pickle.dump(list(X_final.columns), f)
```
